# Parameter Robustness

This notebook checks whether the SPY strategy depends heavily on one lookback or transaction-cost assumption.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from tsmom.data import download_prices
from tsmom.robustness import parameter_sweep, summarize_parameter_grid
from tsmom.plotting import plot_parameter_heatmap
```

```python
prices = download_prices("SPY", start="2000-01-01")
close = prices["SPY"] if "SPY" in prices.columns else prices.iloc[:, 0]
results = parameter_sweep(close, lookbacks=[21, 63, 126, 189, 252], cost_bps_values=[0, 5, 10, 25])
results
```

```python
grid = summarize_parameter_grid(results)
grid
```

```python
plot_parameter_heatmap(results)
```
