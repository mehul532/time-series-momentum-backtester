# SPY Baseline

This notebook walks through a simple 12-month time-series momentum rule on SPY. The signal is shifted by one trading day before returns are applied.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from tsmom.data import download_prices
from tsmom.signals import trailing_return_signal
from tsmom.backtest import backtest_single_asset
from tsmom.metrics import summarize_performance
from tsmom.plotting import plot_equity_curves, plot_drawdowns, plot_signal_over_price
```

```python
prices = download_prices("SPY", start="2000-01-01")
close = prices["SPY"] if "SPY" in prices.columns else prices.iloc[:, 0]
close.tail()
```

```python
signal = trailing_return_signal(close, lookback=252)
results = backtest_single_asset(close, signal, cost_bps=5)
results[["close", "signal", "position", "strategy_return", "benchmark_return"]].tail()
```

```python
summarize_performance(results["strategy_return"], results["benchmark_return"])
```

```python
plot_equity_curves(results)
plot_drawdowns(results)
plot_signal_over_price(close, results["position"])
```
