# Multi-Asset ETF Trend Following

This notebook applies independent time-series momentum signals to a diversified ETF universe. Active ETFs are equal-weighted after signal shifting.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from tsmom.data import download_prices
from tsmom.portfolio import multi_asset_tsmom
from tsmom.metrics import summarize_performance
from tsmom.plotting import plot_equity_curves, plot_multi_asset_exposures
```

```python
universe = ["SPY", "QQQ", "IWM", "TLT", "IEF", "GLD", "DBC", "EFA", "EEM", "VNQ"]
prices = download_prices(universe, start="2005-01-01")
prices.tail()
```

```python
output = multi_asset_tsmom(prices, lookback=252, cost_bps=5)
results = output["results"]
weights = output["weights"]
results.tail()
```

```python
summarize_performance(results["strategy_return"], results["benchmark_return"])
```

```python
plot_equity_curves(results)
plot_multi_asset_exposures(weights)
```
